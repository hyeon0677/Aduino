```python
// C++ code
//

int EN1 = 5;
int EN2 = 6;
int IN1 = 2;
int IN2 = 3;
int IN3 = 7;
int IN4 = 8;

void setup()
{
  pinMode(LED_BUILTIN, OUTPUT);

  pinMode(EN1, OUTPUT);
  pinMode(EN2, OUTPUT);
  pinMode(IN1, OUTPUT);
  pinMode(IN2, OUTPUT);
  pinMode(IN3, OUTPUT);
  pinMode(IN4, OUTPUT);
}

void loop()
{
  digitalWrite(IN1, LOW);
  digitalWrite(IN2, HIGH);  //방향 반대로
  analogWrite(EN1, 1000);   //돌아가는 속도
  delay(1000);

  digitalWrite(IN3, HIGH);
  digitalWrite(IN4, LOW);
  analogWrite(EN2, 1000);
  delay(1000);
}
```
